# NeuroGene Analytics
## Exploratory Brain Gene Expression Analysis of Alzheimer's Disease and Control Hippocampus Samples

**Dataset:** GSE48350 | **Platform:** GPL570

This notebook documents the exploratory bioinformatics and data science workflow. It is not a validated diagnostic or classification model.

## 1. Project Objectives
- Explore brain gene-expression data.
- Perform quality checks and exploratory analysis.
- Investigate differential expression.
- Annotate genes and summarize GO terms.
- Evaluate whether machine-learning classification is scientifically feasible.

## 2. Dataset Information
GEO accession: GSE48350

Platform: GPL570 (Affymetrix Human Genome U133 Plus 2.0).

The hippocampus subset contains 62 samples: 43 controls and 19 AD samples.

## 3. Data Quality and Exploratory Analysis
The analysis checked missing values, duplicate probes, infinite values, expression distributions, sample characteristics and PCA.

## 4. Differential Expression
Welch's t-test was applied to 54,675 probes in the age-restricted 40-sample subset, followed by Benjamini-Hochberg correction.

## 5. Gene Annotation and GO Summary
Probe annotations and GO terms were summarized. GO term frequencies are descriptive and do not represent statistical enrichment.

## 6. Visualizations
Saved outputs include the volcano plot, PCA plot and top-20 gene heatmap.

## 7. Machine Learning Feasibility
Disease status is perfectly confounded with accession group in the age-restricted subset. Therefore, no classifier was trained or reported.

## 8. Limitations and Conclusion
Observed expression differences cannot be attributed specifically to Alzheimer's disease because cohort/accession/processing confounding remains unresolved. Independent validation data are needed.
